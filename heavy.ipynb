```python
import random

def flip_coin():
    """
    Simulates a weighted coin flip.
    Returns 1 for heads (1% probability) or 0 for tails (99% probability).
    """
    return 1 if random.random() < 0.01 else 0

# Test the function
print("Single flip:", flip_coin())

# Test with multiple flips to see the distribution
flips = [flip_coin() for _ in range(1000)]
heads_count = sum(flips)
tails_count = len(flips) - heads_count
print(f"\nOut of 1000 flips:")
print(f"Heads (1): {heads_count} ({heads_count/10:.1f}%)")
print(f"Tails (0): {tails_count} ({tails_count/10:.1f}%)")
```

```python
def flip_x_times(x):
    """
    Simulates flipping a weighted coin x times.
    Returns a list of 1s and 0s representing heads and tails.
    """
    return [flip_coin() for _ in range(x)]

outcomes_with_at_least_one_head = []
outcomes_with_no_heads = []
trials = 100000

for _ in range(trials):
    flips = flip_x_times(1000)
    heads_count = sum(flips)
    if heads_count == 0:
        outcomes_with_no_heads.append(flips)

print(len(outcomes_with_no_heads))
print(f"{len(outcomes_with_no_heads)/trials * 100:.4f}%")
```
